```python
#It includes Puerto Rico 
print(df["latitude"].median())
print(df["longitude"].median())
```

```python
df["abslat"] = (df["latitude"] - 39.4355155).abs()
print(df["abslat"])
```

```python
df["abslon"] = (df["longitude"] - (-89.093)).abs()
print(df["abslon"])
```

```python
print(df.shape)
```

```python
row = df["abslat"].idxmin()
print(f"{df.loc[row, "abslat"]:.2f}")

```

```python
row1 = df["abslat"].idxmax()
print(f"{df.loc[row1, "abslat"]:.2f}")
```

```python
row2 = df["abslon"].idxmin()
print(f"{df.loc[row2, "abslon"]:.2f}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("planets")
```

```python
df = df.dropna(axis = 1)
```

```python
print(df.shape)
```

```python
print(df.head())
print(df.tail())
```

```python
fig, axes = plt.subplots(2,2)
axes[0,0].scatter(df["year"], df["number"])
axes[0,1].scatter(df["number"], df["year"], color = "green")
axes[1,0].boxplot(df["year"]) 
sns.histplot(df["year"]), axes[1,1]
axes[1,0].set_title("kaboom")
axes[1,1].set_title("kablam")
axes[0,0].set_title("kerchow")
axes[0,1].set_title("sdifub")

axes[1,0].set_xlabel("kaboom")
axes[1,1].set_title("kablam")
axes[0,0].set_title("kerchow")
axes[0,1].set_title("sdifub")

```
