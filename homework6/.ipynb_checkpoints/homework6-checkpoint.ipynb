```python
import numpy as np 
```

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)

print(arr * 4)
```

```python
print(np.mean(arr))
```

```python
print(np.std(arr))
```

```python
print(np.pargmin(arr))
print(np.argmax(arr))
```

```python
print(arr[6])
print(arr[7])
```

```python
print(np.min(arr))
print(np.max(arr))
```

```python
print(arr[arr > 20])
```

```python
matrix = [[5, 12, 17, 9], [ 8, 21, 14, 3], [6, 13, 19, 11]]
print(np.shape(matrix))
```

```python
# 3 rows 4 columns 
matrix = np.array(matrix)
print(matrix[:,2])
```

```python
print(matrix[1,:])
```

```python
print(matrix * 2)
```

```python
n = len(matrix[0])
for x in range(n):
    print(np.mean(matrix[:,x]))
```

```python
print(matrix.mean(axis=0))
```

```python
print(matrix.mean())
```

```python
print(matrix.max(axis=1))
print(matrix.min(axis=0))
```

```python
print(matrix.flatten())
yarr = matrix.flatten()
```

```python
print(np.argmin(yarr), np.argmax(yarr))
```

```python
print(np.unravel_index(7, matrix.shape))
```

```python
print(np.unravel_index(5, matrix.shape))
```

```python
print(np.matrix[matrix > 15])
```
