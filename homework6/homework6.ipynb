```python
stel = np.array([[2,3,5],[14,6,8],[11,13,17],[7,10,13]])
```

```python
def isprime(x):
    if x < 2: 
        return False
    for d in range (2, x):
        if x % d == 0:
            return False 
    else:
        return True 
```

```python
def primerow(ray):
    for row in ray: 
        for n in row:
            if isprime(n) == True:
                print(row)
                break
primerow(stel)
                
```

```python
def expansion(string, num): 
    spacer = " " * num 
    ufo = spacer.join(string)
    return ufo 

print(expansion("bazinga", 3))
    
```
