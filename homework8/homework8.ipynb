```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    y = A * np.sin(w * x)
    return y
x = np.linspace(0,2*np.pi,1000)
y = (make_sine_wave(x, 3, 3))

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1,2,3,4,5]

fig, axes = plt.subplots(3, 1, figsize = (25,20))
for a, w in zip(amplitude, frequency):
    axes[0].plot(x, make_sine_wave(x, a, 1), label = f"A={a}")
    axes[1].plot(x, make_sine_wave(x, 1, w), label = f"w={w}")
    axes[2].plot(x, make_sine_wave(x, a, w), label = f"A={a}")
    axes[0].set_title("Amplitude Varies", fontsize = 40)
    axes[1].set_title("Frequency Varies", fontsize = 40)
    axes[2].set_title("Both Vary", fontsize = 40)
    for i in range(0,3):
        axes[i].set_xlabel("x", fontsize = 20)
        axes[i].set_ylabel("y", fontsize = 20)
        axes[i].legend(fontsize = 35)
plt.show()




                
```
