```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
data = np.loadtxt("../../course_assignments/homework8/star_data.txt", skiprows=1)
```

```python
data.shape
```

```python
magnitude = data[:,0]
temperature = data[:,1]
```

```python
print(magnitude.min())
print(magnitude.max())
print(temperature.min())
print(temperature.max())
```

```python
print(magnitude.argmin())
print(magnitude.argmax())
print(temperature.argmin())
print(temperature.argmax())
```

```python
#1988 is the brightest, 264 is the dimmest 
```

```python
log_mag = np.log(magnitude)
log_temp= np.log(temperature)

fig, axes = plt.subplots(1,1, figsize=(8,6))
axes.scatter(log_temp, log_mag)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

brightest = (log_temp <8.4) & (log_mag <1) 
axes.scatter(log_temp[brightest], log_mag[brightest], s = 150, color = "black", label = "Radiohead", marker = "*")
dimmest = (log_temp >8.95) & (log_mag > 2.25)
axes.scatter(log_temp[dimmest], log_mag[dimmest], s = 150, color = "purple", label = "Dimmest", marker = "*")
hottest = np.argmax(log_temp)
axes.scatter(log_temp[hottest], log_mag[hottest], s = 150, color = "green", marker = "*", label = "hottest")
coldest = np.argmin(log_temp)
axes.scatter(log_temp[coldest], log_mag[coldest], s = 300, color = "green", marker = "*", label = "coldest")
axes.set_title("Hertzsprung-Russel Diagram")
axes.set_xlabel("Temperature (log K)")
axes.set_ylabel("Magnitude")
plt.legend()
```
